2. Tool: Get Stock Price

```python
def get_stock_price(symbol: str) -> dict:
    """
    Get latest stock price using Financial Modeling Prep API
    """
    try:
        url = f"https://financialmodelingprep.com/api/v3/quote/{symbol.upper()}?apikey=demo"
        response = requests.get(url)
        data = response.json()

        if not data:
            return {"error": "Stock not found"}

        stock = data[0]

        return {
            "symbol": stock["symbol"],
            "price": stock["price"],
            "day_high": stock["dayHigh"],
            "day_low": stock["dayLow"],
            "volume": stock["volume"]
        }

    except Exception as e:
        return {"error": str(e)}
```

3. Tool: Calculator

```python
def calculator(expression: str) -> dict:
    """
    Simple calculator tool
    """
    try:
        result = eval(expression, {"__builtins__": {}})
        return {"expression": expression, "result": result}
    except:
        return {"error": "Invalid math expression"}
```

4. Gemini Agent Configuration

```python
system_prompt = """
You are a financial assistant AI agent.

You can:
1. Look up stock prices
2. Perform calculations

Rules:
- Use tools when needed
- Explain results clearly
- Do not provide financial advice
"""

config = types.GenerateContentConfig(
    system_instruction=system_prompt,
    tools=[get_stock_price, calculator]
)

```

5. Test Example

```python
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="What is the current stock price of Apple (AAPL)?",
    config=config
)

print("TEST RESULT:")
print(response.text)
```

 6. Chat Loop

```python
print("\n=== Stock Market AI Agent ===")
print("Example questions:")
print("What is the stock price of Tesla?")
print("Compare AAPL and MSFT")
print("What is 120 * 35?")
print("Type 'exit' to quit\n")

while True:

    user = input("You: ")

    if user.lower() == "exit":
        break

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=user,
        config=config
    )

    print("Agent:", response.text)
```
