```python
import os
from google import genai

# 1. Setup
client = genai.Client(api_key="AIzaSyBFw6spBYdwjLLW-LhQmVdXa60FO-bkGkQ")

# 2. Initialize the Model with a System Instruction
# The system instruction sets the behavior of the "Agent"
model_id = "gemini-2.0-flash"
system_prompt = "You are a witty science tutor. Keep explanations simple but insightful."

# 3. Start a Chat Session (This handles the 'Memory')
chat = client.chats.create(
    model=model_id,
    config={'system_instruction': system_prompt}
)

print("--- Gemini Chatbot Active (type 'exit' to stop) ---")

# 4. The Chat Loop
while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit", "bye"]:
        print("Bot: Goodbye!")
        break

    # send_message appends the prompt to history and gets the response
    response = chat.send_message(user_input)

    print(f"Agent: {response.text}\n")
```
